```python
# ============================================
# LIBRERÍAS REQUERIDAS
# ============================================
import pandas as pd
import matplotlib.pyplot as plt
import glob
import random
```

```python
# ============================================
# 1. CONVERTIR EXCEL A CSV
# ============================================
# Leer archivo Excel y guardarlo en formato CSV
df = pd.read_excel("./data/online+retail/Online Retail.xlsx")
df.to_csv("./data/online+retail/onlineRetail.csv", index=False)

print(f"\n=== CONVERSIÓN COMPLETADA ===\n")
print(f"Registros convertidos: {len(df):,}")
```

```python
# ============================================
# 2. DIVIDIR ARCHIVO NOVIEMBRE
# ============================================
# Partir archivo grande en chunks de 1 millón de registros
chunk_size = 1000000

input_file = pd.read_csv("./data/online+retail/2019-Nov.csv", chunksize=chunk_size)

for i, chunk in enumerate(input_file):
    output_file = f"./data/online+retail/split/2019-Nov_part_{i}.csv"
    chunk.to_csv(output_file, index=False)

print(f"\n=== DIVISIÓN NOVIEMBRE COMPLETADA ===\n")
```

```python
# ============================================
# 3. DIVIDIR ARCHIVO OCTUBRE
# ============================================
# Partir archivo octubre en chunks de 1 millón de registros
input_file = pd.read_csv("./data/online+retail/2019-Oct.csv", chunksize=chunk_size)

for i, chunk in enumerate(input_file):
    output_file = f"./data/online+retail/split/2019-Oct_part_{i}.csv"
    chunk.to_csv(output_file, index=False)

print(f"\n=== DIVISIÓN OCTUBRE COMPLETADA ===\n")
```
